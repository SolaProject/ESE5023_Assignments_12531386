```python
import geopandas as gpd
import regionmask
import netCDF4 as nc
```

```python
gdf_world = gpd.read_file("./data/world.zh.json")
gdf_china = gpd.read_file("./data/china_province.geojson").sort_values(by="gb")
gdf_china = gdf_china.loc[gdf_china["name"]!="境界线"].reset_index()
with nc.Dataset("./data/HAQ_TROPOMI_NO2_GLOBAL_QA75_L3_Monthly_Mean_V2.4_201901_to_202508.nc") as nf:
    lon, lat = nf["lon"][:].data, nf["lat"][:].data
    mask = regionmask.mask_geopandas(gdf_world, lon, lat)
    # setattr(mask, "name", gdf_world["name"].to_list())
    mask.to_netcdf("./data/mask_world_country.nc", encoding={"mask": {"zlib": True}})
with nc.Dataset("./data/mask_world_country.nc", "r+") as nf:
    nf["mask"].setncattr("name", gdf_world["name"].to_list())
with nc.Dataset("./data/HAQ_TROPOMI_NO2_CHINA_QA75_L3_Monthly_V2.4.nc") as nf:
    lon, lat = nf["lon"][:].data, nf["lat"][:].data
    mask = regionmask.mask_geopandas(gdf_china, lon, lat)
    # setattr(mask, "name", gdf_china["name"].to_list())
    mask.to_netcdf("./data/mask_china_province.nc", encoding={"mask": {"zlib": True}})
with nc.Dataset("./data/mask_china_province.nc", "r+") as nf:
    nf["mask"].setncattr("name", gdf_china["name"].to_list())
```
